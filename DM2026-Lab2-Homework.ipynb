### 1.1 Preprocessing Steps

I used 3,360 labeled posts and 1,440 test posts with six emotion classes. I kept the 50 repeated training texts. ### 1.2 Feature Engineering Steps

I started with word TF-IDF, which produced 6,541 features. Adding character n-grams of length 3–5 increased this to 29,901 features. These can capture parts of words and spelling variations. The vectorizers were fitted only on the training part. I also tested frozen MPNet embeddings and fine-tuned DistilRoBERTa, RoBERTa-base and DeBERTa-v3-base. All models used the same validation split.

The basic SVM reached a validation macro F1 of 0.4711. Class balancing, character features and lowering C to 0.3 improved it to 0.5162. MPNet was worse at 0.4593. ## 1. Preprocessing Steps

```python
from pathlib import Path
from zipfile import ZipFile
import json
import pandas as pd

zip_path = Path.home() / "Downloads" / "2026-dm-lab-2-private-competition.zip"

with ZipFile(zip_path) as archive:
    posts_df = pd.DataFrame(json.loads(archive.read("final_posts.json")))
    split_df = pd.read_csv(
        archive.open("data_identification.csv"), dtype={"post_id": str}
    )
    labels_df = pd.read_csv(
        archive.open("emotion.csv"), dtype={"post_id": str}
    )
    sample_submission = pd.read_csv(
        archive.open("sampleSubmission.csv"), dtype={"post_id": str}
    )

dataset = posts_df.merge(split_df, on="post_id", validate="one_to_one")
dataset = dataset.merge(
    labels_df, on="post_id", how="left", validate="one_to_one"
)

train_df = dataset.loc[
    dataset["set"] == "train", ["post_id", "text", "label"]
].reset_index(drop=True)

test_df = dataset.loc[
    dataset["set"] == "test", ["post_id", "text"]
].reset_index(drop=True)

assert train_df["label"].notna().all()
assert set(test_df["post_id"]) == set(sample_submission["post_id"])

print("Training shape:", train_df.shape)
print("Test shape:", test_df.shape)
print("\nTraining labels:")
print(train_df["label"].value_counts())
print("\nRepeated training texts:", train_df["text"].duplicated().sum())
print("Submission columns:", sample_submission.columns.tolist())

from sklearn.model_selection import StratifiedGroupKFold

SEED = 42

text_groups = (
    train_df["text"].str.casefold()
    .str.replace(r"\s+", " ", regex=True).str.strip()
)

splitter = StratifiedGroupKFold(
    n_splits=5, shuffle=True, random_state=SEED
)

train_idx, valid_idx = next(
    splitter.split(train_df["text"], train_df["label"], groups=text_groups)
)

train_part = train_df.iloc[train_idx].copy()
valid_part = train_df.iloc[valid_idx].copy()

shared_groups = set(text_groups.iloc[train_idx]) & set(text_groups.iloc[valid_idx])
assert not shared_groups

print("Training rows:", len(train_part))
print("Validation rows:", len(valid_part))
print("Shared text groups:", len(shared_groups))

display(pd.DataFrame({
    "training": train_part["label"].value_counts(),
    "validation": valid_part["label"].value_counts()
}).sort_index())
```

## 2. Feature Engineering Steps

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(ngram_range=(1, 1))

X_train_tfidf = vectorizer.fit_transform(train_part["text"])
X_valid_tfidf = vectorizer.transform(valid_part["text"])

y_train = train_part["label"]
y_valid = valid_part["label"]

print("Training matrix:", X_train_tfidf.shape)
print("Validation matrix:", X_valid_tfidf.shape)
print("Vocabulary size:", len(vectorizer.get_feature_names_out()))

from sklearn.pipeline import FeatureUnion

combined_vectorizer = FeatureUnion([
    ("words", TfidfVectorizer(ngram_range=(1, 1))),
    ("characters", TfidfVectorizer(
        analyzer="char_wb", ngram_range=(3, 5), min_df=2
    ))
])

X_train_combined = combined_vectorizer.fit_transform(train_part["text"])
X_valid_combined = combined_vectorizer.transform(valid_part["text"])

print("Combined training matrix:", X_train_combined.shape)
print("Combined validation matrix:", X_valid_combined.shape)
```

## 3. Model Implementation Steps

```python
from sklearn.svm import LinearSVC
from sklearn.metrics import f1_score, classification_report

baseline_model = LinearSVC(
    C=1.0, random_state=SEED, max_iter=5000
)

baseline_model.fit(X_train_tfidf, y_train)
baseline_predictions = baseline_model.predict(X_valid_tfidf)

baseline_macro_f1 = f1_score(
    y_valid, baseline_predictions, average="macro"
)

print(f"Validation macro F1: {baseline_macro_f1:.4f}")
print(classification_report(
    y_valid, baseline_predictions, digits=3, zero_division=0
))

balanced_model = LinearSVC(
    C=1.0,
    class_weight="balanced",
    random_state=SEED,
    max_iter=5000
)

balanced_model.fit(X_train_tfidf, y_train)
balanced_predictions = balanced_model.predict(X_valid_tfidf)

balanced_macro_f1 = f1_score(
    y_valid, balanced_predictions, average="macro"
)

print(f"Baseline macro F1: {baseline_macro_f1:.4f}")
print(f"Balanced macro F1: {balanced_macro_f1:.4f}")
print(classification_report(
    y_valid, balanced_predictions, digits=3, zero_division=0
))

combined_model = LinearSVC(
    C=1.0, class_weight="balanced", random_state=SEED, max_iter=5000
)

combined_model.fit(X_train_combined, y_train)
combined_predictions = combined_model.predict(X_valid_combined)

combined_macro_f1 = f1_score(
    y_valid, combined_predictions, average="macro"
)

print(f"Baseline: {baseline_macro_f1:.4f}")
print(f"Balanced: {balanced_macro_f1:.4f}")
print(f"Words + characters: {combined_macro_f1:.4f}")
print(classification_report(
    y_valid, combined_predictions, digits=3, zero_division=0
))

c_models = {1.0: combined_model}
c_scores = {1.0: combined_macro_f1}

for c_value in [0.3, 3.0]:
    trial_model = LinearSVC(
        C=c_value,
        class_weight="balanced",
        random_state=SEED,
        max_iter=5000
    )
    trial_model.fit(X_train_combined, y_train)
    trial_predictions = trial_model.predict(X_valid_combined)

    c_models[c_value] = trial_model
    c_scores[c_value] = f1_score(
        y_valid, trial_predictions, average="macro"
    )

display(pd.DataFrame({
    "C": list(c_scores.keys()),
    "macro_f1": list(c_scores.values())
}).sort_values("macro_f1", ascending=False).round(4))

best_C = max(c_scores, key=c_scores.get)
best_model = c_models[best_C]

print(f"Best C: {best_C}")
print(f"Best validation macro F1: {c_scores[best_C]:.4f}")

from sklearn.base import clone

final_vectorizer = clone(combined_vectorizer)
X_full = final_vectorizer.fit_transform(train_df["text"])

test_ordered = test_df.set_index("post_id").loc[
    sample_submission["post_id"]
]
X_test_final = final_vectorizer.transform(test_ordered["text"])

final_model = clone(best_model)
final_model.fit(X_full, train_df["label"])

submission = sample_submission.copy()
submission["label"] = final_model.predict(X_test_final)

assert len(submission) == len(test_df)
assert submission["post_id"].is_unique
assert submission.notna().all().all()

submission_kaggle = submission.rename(columns={"post_id": "id"})
assert submission_kaggle.columns.tolist() == ["id", "label"]

submission_path = Path.home() / "Downloads" / "submission_svm_word_char_id.csv"
submission_kaggle.to_csv(submission_path, index=False)

print("Saved:", submission_path)
display(submission_kaggle.head())
```

```python
# Fine-tune a general masked-language model using the existing grouped holdout.
import gc
import math
import platform
import random
import re
import time

import sklearn
import torch
import transformers
from sklearn.metrics import accuracy_score, classification_report, f1_score
from torch.utils.data import DataLoader, Dataset
from transformers import (
    AutoModelForSequenceClassification,
    AutoTokenizer,
    get_linear_schedule_with_warmup,
)


DISTIL_MODEL_ID = "distilbert/distilroberta-base"
if not isinstance(globals().get("DISTIL_REVISION"), str) or not re.fullmatch(
    r"[0-9a-fA-F]{40}", DISTIL_REVISION
):
    raise ValueError(
        "Set DISTIL_REVISION to the exact commit SHA returned by "
        "HfApi().model_info('distilbert/distilroberta-base').sha before running."
    )


def train_distilroberta(
    training_frame,
    validation_frame=None,
    *,
    revision,
    output_dir,
    epochs=3,
    seed=42,
):
    """Train from the pinned general checkpoint; return metadata and holdout labels.

    For a selected final refit, pass all training rows, validation_frame=None,
    epochs=distil_result['best_epoch'], and a separate output_dir. This trains a
    fresh model for that fixed number of epochs and saves its final checkpoint.
    """
    if not re.fullmatch(r"[0-9a-fA-F]{40}", str(revision)):
        raise ValueError("revision must be an immutable Hugging Face commit SHA")
    if not isinstance(epochs, int) or not 1 <= epochs <= 3:
        raise ValueError("epochs must be an integer between 1 and 3")
    if training_frame.empty or training_frame[["text", "label"]].isna().any().any():
        raise ValueError("training_frame must contain non-null text and labels")
    if validation_frame is not None:
        if validation_frame.empty or validation_frame[["text", "label"]].isna().any().any():
            raise ValueError("validation_frame must contain non-null text and labels")

    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
    if device.type == "mps":
        torch.mps.manual_seed(seed)

    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    cache_dir = experiment_dir / "hf_cache"
    cache_dir.mkdir(parents=True, exist_ok=True)
    labels = sorted(training_frame["label"].astype(str).unique().tolist())
    if len(labels) != 6:
        raise ValueError(f"Expected six training classes, found {labels}")
    label_to_id = {label: index for index, label in enumerate(labels)}
    id_to_label = {index: label for label, index in label_to_id.items()}
    train_targets = training_frame["label"].astype(str).map(label_to_id).to_numpy()
    valid_targets = None
    if validation_frame is not None:
        mapped = validation_frame["label"].astype(str).map(label_to_id)
        if mapped.isna().any():
            raise ValueError("Validation includes a class absent from training")
        valid_targets = mapped.to_numpy(dtype=np.int64)

    # All class weights are derived exclusively from this call's training rows.
    class_counts = np.bincount(train_targets, minlength=len(labels))
    class_weights = len(train_targets) / (len(labels) * class_counts)
    settings = {
        "seed": int(seed),
        "num_labels": 6,
        "max_length": 128,
        "batch_size": 8,
        "gradient_accumulation_steps": 2,
        "learning_rate": 2e-5,
        "weight_decay": 0.01,
        "warmup_ratio": 0.10,
        "epochs_requested": epochs,
        "max_epochs": 3,
        "optimizer": "torch.optim.AdamW",
        "adam_betas": [0.9, 0.999],
        "adam_epsilon": 1e-8,
        "scheduler": "linear warmup followed by linear decay",
        "max_grad_norm": 1.0,
        "dtype": "float32",
        "amp": False,
        "num_workers": 0,
        "padding": "max_length",
        "selection_metric": "validation macro F1" if validation_frame is not None else None,
        "selection_rule": "highest macro F1; earliest epoch wins ties" if validation_frame is not None else "fixed final epoch",
        "class_weight_formula": "n_training_rows / (n_classes * training_class_count)",
        "class_counts": {label: int(class_counts[index]) for index, label in enumerate(labels)},
        "class_weights": {label: float(class_weights[index]) for index, label in enumerate(labels)},
        "trust_remote_code": False,
        "use_safetensors": True,
    }
    versions = {
        "python": platform.python_version(),
        "platform": platform.platform(),
        "torch": str(torch.__version__),
        "transformers": transformers.__version__,
        "sklearn": sklearn.__version__,
        "numpy": np.__version__,
        "pandas": pd.__version__,
    }
    result = {
        "best_macro_f1": None,
        "best_epoch": None,
        "model_id": DISTIL_MODEL_ID,
        "revision": revision,
        "device": str(device),
        "settings": settings,
        "versions": versions,
        "label_classes": labels,
        "train_rows": int(len(training_frame)),
        "validation_rows": int(len(validation_frame)) if validation_frame is not None else 0,
        "checkpoint_dir": str(output_dir.resolve()),
        "epoch_history": [],
        "seed_note": "Python, NumPy, Torch, MPS and DataLoader seeded; backend operations may remain nondeterministic.",
    }

    tokenizer = AutoTokenizer.from_pretrained(
        DISTIL_MODEL_ID,
        revision=revision,
        cache_dir=str(cache_dir),
        token=False,
        trust_remote_code=False,
    )

    class EncodedTextDataset(Dataset):
        def __init__(self, frame, targets):
            self.features = tokenizer(
                frame["text"].astype(str).tolist(),
                truncation=True,
                padding="max_length",
                max_length=settings["max_length"],
                return_token_type_ids=False,
                return_tensors="pt",
            )
            self.targets = torch.tensor(np.array(targets, copy=True), dtype=torch.long)

        def __len__(self):
            return len(self.targets)

        def __getitem__(self, index):
            return {
                **{key: value[index] for key, value in self.features.items()},
                "labels": self.targets[index],
            }

    loader_generator = torch.Generator().manual_seed(seed)
    train_loader = DataLoader(
        EncodedTextDataset(training_frame, train_targets),
        batch_size=settings["batch_size"],
        shuffle=True,
        generator=loader_generator,
        num_workers=0,
        pin_memory=False,
    )
    valid_loader = None
    if validation_frame is not None:
        valid_loader = DataLoader(
            EncodedTextDataset(validation_frame, valid_targets),
            batch_size=settings["batch_size"],
            shuffle=False,
            num_workers=0,
            pin_memory=False,
        )

    accumulation = settings["gradient_accumulation_steps"]
    updates_per_epoch = math.ceil(len(train_loader) / accumulation)
    total_updates = updates_per_epoch * epochs
    warmup_updates = math.ceil(settings["warmup_ratio"] * total_updates)
    settings.update({
        "optimizer_updates_per_epoch": updates_per_epoch,
        "total_optimizer_updates": total_updates,
        "warmup_optimizer_updates": warmup_updates,
    })
    print(json.dumps(result, indent=2), flush=True)

    model = optimizer = scheduler = criterion = None
    best_predictions = None
    best_score = -float("inf")
    global_update = 0
    started = time.monotonic()
    try:
        # Loading the general MLM checkpoint creates a fresh six-class head.
        model = AutoModelForSequenceClassification.from_pretrained(
            DISTIL_MODEL_ID,
            revision=revision,
            cache_dir=str(cache_dir),
            token=False,
            num_labels=6,
            id2label=id_to_label,
            label2id=label_to_id,
            trust_remote_code=False,
            use_safetensors=True,
        ).float().to(device)
        criterion = torch.nn.CrossEntropyLoss(
            weight=torch.as_tensor(class_weights, dtype=torch.float32, device=device)
        )
        optimizer = torch.optim.AdamW(
            model.parameters(),
            lr=settings["learning_rate"],
            betas=tuple(settings["adam_betas"]),
            eps=settings["adam_epsilon"],
            weight_decay=settings["weight_decay"],
        )
        scheduler = get_linear_schedule_with_warmup(
            optimizer,
            num_warmup_steps=warmup_updates,
            num_training_steps=total_updates,
        )

        for epoch in range(1, epochs + 1):
            model.train()
            optimizer.zero_grad(set_to_none=True)
            training_loss_sum = 0.0
            training_weight_sum = 0.0
            for batch_index, batch in enumerate(train_loader):
                targets = batch.pop("labels").to(device)
                inputs = {key: value.to(device) for key, value in batch.items()}
                logits = model(**inputs).logits
                loss = criterion(logits, targets)
                batch_weight = float(criterion.weight[targets].sum().item())
                training_loss_sum += float(loss.detach().item()) * batch_weight
                training_weight_sum += batch_weight
                # Scale the final accumulation group correctly if it is shorter.
                group_start = (batch_index // accumulation) * accumulation
                group_size = min(accumulation, len(train_loader) - group_start)
                (loss / group_size).backward()
                if (batch_index + 1) % accumulation == 0 or batch_index + 1 == len(train_loader):
                    torch.nn.utils.clip_grad_norm_(model.parameters(), settings["max_grad_norm"])
                    optimizer.step()
                    scheduler.step()
                    optimizer.zero_grad(set_to_none=True)
                    global_update += 1
                    if global_update % 25 == 0 or global_update == total_updates:
                        print(
                            f"epoch {epoch}/{epochs} | optimizer step {global_update}/{total_updates} "
                            f"| train loss {training_loss_sum / training_weight_sum:.4f} "
                            f"| elapsed {time.monotonic() - started:.1f}s",
                            flush=True,
                        )
                del logits, loss, targets, inputs

            epoch_record = {
                "epoch": epoch,
                "train_loss": training_loss_sum / training_weight_sum,
                "optimizer_steps_completed": global_update,
                "elapsed_seconds": time.monotonic() - started,
            }
            predictions = None
            if valid_loader is not None:
                model.eval()
                predicted_ids = []
                with torch.inference_mode():
                    for batch in valid_loader:
                        batch.pop("labels")
                        inputs = {key: value.to(device) for key, value in batch.items()}
                        logits = model(**inputs).logits
                        predicted_ids.extend(logits.argmax(dim=-1).cpu().tolist())
                        del logits, inputs
                predicted_ids = np.asarray(predicted_ids, dtype=np.int64)
                predictions = np.asarray([id_to_label[int(index)] for index in predicted_ids])
                score = float(f1_score(
                    valid_targets, predicted_ids, labels=list(range(6)),
                    average="macro", zero_division=0,
                ))
                epoch_record.update({
                    "validation_macro_f1": score,
                    "validation_accuracy": float(accuracy_score(valid_targets, predicted_ids)),
                })
                improved = score > best_score
            else:
                score = None
                improved = epoch == epochs

            result["epoch_history"].append(epoch_record)
            print(json.dumps(epoch_record, indent=2), flush=True)
            if improved:
                if score is not None:
                    best_score = score
                result["best_macro_f1"] = score
                result["best_epoch"] = epoch
                best_predictions = predictions.copy() if predictions is not None else None
                model.save_pretrained(str(output_dir), safe_serialization=True)
                tokenizer.save_pretrained(str(output_dir))
                if best_predictions is not None:
                    pd.DataFrame({
                        "post_id": validation_frame["post_id"].to_numpy(),
                        "true_label": validation_frame["label"].astype(str).to_numpy(),
                        "predicted_label": best_predictions,
                    }).to_csv(output_dir / "validation_predictions.csv", index=False)
                print(f"Saved checkpoint at epoch {epoch} to {output_dir}", flush=True)
            (output_dir / "epoch_results.json").write_text(
                json.dumps(result, indent=2), encoding="utf-8"
            )

        print("Epoch scores:", flush=True)
        print(pd.DataFrame(result["epoch_history"]).to_string(index=False), flush=True)
        if best_predictions is not None:
            report = classification_report(
                validation_frame["label"].astype(str).to_numpy(),
                best_predictions,
                labels=labels,
                digits=4,
                zero_division=0,
            )
            result["best_classification_report"] = classification_report(
                validation_frame["label"].astype(str).to_numpy(),
                best_predictions,
                labels=labels,
                output_dict=True,
                zero_division=0,
            )
            print(f"Best epoch: {result['best_epoch']} | macro F1: {result['best_macro_f1']:.6f}", flush=True)
            print(report, flush=True)
        result["total_elapsed_seconds"] = time.monotonic() - started
        (output_dir / "epoch_results.json").write_text(
            json.dumps(result, indent=2), encoding="utf-8"
        )
        return result, best_predictions
    finally:
        # Return only metadata/predictions so later experiments can reclaim memory.
        model = optimizer = scheduler = criterion = None
        gc.collect()
        if device.type == "mps":
            torch.mps.empty_cache()


distil_result, distil_validation_predictions = train_distilroberta(
    train_part,
    valid_part,
    revision=DISTIL_REVISION,
    output_dir=experiment_dir / "distilroberta_best",
    epochs=3,
    seed=SEED,
)
```

```python
# Larger base-model comparison; existing runs and checkpoints are retained.
# Fine-tune a general masked-language model using the existing grouped holdout.
import gc
import math
import platform
import random
import re
import time

import sklearn
import torch
import transformers
from sklearn.metrics import accuracy_score, classification_report, f1_score
from torch.utils.data import DataLoader, Dataset
from transformers import (
    AutoModelForSequenceClassification,
    AutoTokenizer,
    get_linear_schedule_with_warmup,
)


def train_candidate(
    training_frame,
    validation_frame=None,
    *,
    model_id,
    revision,
    output_dir,
    epochs=3,
    seed=42,
    learning_rate=2e-5,
    weight_power=1.0,
):
    """Fit a pinned base model; save the best validation checkpoint."""
    if not re.fullmatch(r"[0-9a-fA-F]{40}", str(revision)):
        raise ValueError("revision must be an immutable Hugging Face commit SHA")
    if not isinstance(epochs, int) or not 1 <= epochs <= 3:
        raise ValueError("epochs must be an integer between 1 and 3")
    if training_frame.empty or training_frame[["text", "label"]].isna().any().any():
        raise ValueError("training_frame must contain non-null text and labels")
    if validation_frame is not None:
        if validation_frame.empty or validation_frame[["text", "label"]].isna().any().any():
            raise ValueError("validation_frame must contain non-null text and labels")

    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
    if device.type == "mps":
        torch.mps.manual_seed(seed)

    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    cache_dir = experiment_dir / "hf_cache"
    cache_dir.mkdir(parents=True, exist_ok=True)
    labels = sorted(training_frame["label"].astype(str).unique().tolist())
    if len(labels) != 6:
        raise ValueError(f"Expected six training classes, found {labels}")
    label_to_id = {label: index for index, label in enumerate(labels)}
    id_to_label = {index: label for label, index in label_to_id.items()}
    train_targets = training_frame["label"].astype(str).map(label_to_id).to_numpy()
    valid_targets = None
    if validation_frame is not None:
        mapped = validation_frame["label"].astype(str).map(label_to_id)
        if mapped.isna().any():
            raise ValueError("Validation includes a class absent from training")
        valid_targets = mapped.to_numpy(dtype=np.int64)

    # All class weights are derived exclusively from this call's training rows.
    class_counts = np.bincount(train_targets, minlength=len(labels))
    class_weights = (len(train_targets) / (len(labels) * class_counts)) ** weight_power
    settings = {
        "seed": int(seed),
        "num_labels": 6,
        "max_length": 128,
        "batch_size": 8,
        "gradient_accumulation_steps": 2,
        "learning_rate": learning_rate,
        "weight_decay": 0.01,
        "warmup_ratio": 0.10,
        "epochs_requested": epochs,
        "max_epochs": 3,
        "optimizer": "torch.optim.AdamW",
        "adam_betas": [0.9, 0.999],
        "adam_epsilon": 1e-8,
        "scheduler": "linear warmup followed by linear decay",
        "max_grad_norm": 1.0,
        "dtype": "float32",
        "amp": False,
        "num_workers": 0,
        "padding": "max_length",
        "selection_metric": "validation macro F1" if validation_frame is not None else None,
        "selection_rule": "highest macro F1; earliest epoch wins ties" if validation_frame is not None else "fixed final epoch",
        "class_weight_formula": "(n_training_rows / (n_classes * training_class_count)) ** weight_power",
        "weight_power": weight_power,
        "loss_normalization": "mean of weighted microbatch means (same as original baseline)",
        "class_counts": {label: int(class_counts[index]) for index, label in enumerate(labels)},
        "class_weights": {label: float(class_weights[index]) for index, label in enumerate(labels)},
        "trust_remote_code": False,
        "use_safetensors": True,
    }
    versions = {
        "python": platform.python_version(),
        "platform": platform.platform(),
        "torch": str(torch.__version__),
        "transformers": transformers.__version__,
        "sklearn": sklearn.__version__,
        "numpy": np.__version__,
        "pandas": pd.__version__,
    }
    result = {
        "best_macro_f1": None,
        "best_epoch": None,
        "model_id": model_id,
        "revision": revision,
        "device": str(device),
        "settings": settings,
        "versions": versions,
        "label_classes": labels,
        "train_rows": int(len(training_frame)),
        "validation_rows": int(len(validation_frame)) if validation_frame is not None else 0,
        "checkpoint_dir": str(output_dir.resolve()),
        "epoch_history": [],
        "seed_note": "Python, NumPy, Torch, MPS and DataLoader seeded; backend operations may remain nondeterministic.",
    }

    tokenizer = AutoTokenizer.from_pretrained(
        model_id,
        revision=revision,
        cache_dir=str(cache_dir),
        token=False,
        trust_remote_code=False,
    )

    class EncodedTextDataset(Dataset):
        def __init__(self, frame, targets):
            self.features = tokenizer(
                frame["text"].astype(str).tolist(),
                truncation=True,
                padding="max_length",
                max_length=settings["max_length"],
                return_token_type_ids=False,
                return_tensors="pt",
            )
            self.targets = torch.tensor(np.array(targets, copy=True), dtype=torch.long)

        def __len__(self):
            return len(self.targets)

        def __getitem__(self, index):
            return {
                **{key: value[index] for key, value in self.features.items()},
                "labels": self.targets[index],
            }

    loader_generator = torch.Generator().manual_seed(seed)
    train_loader = DataLoader(
        EncodedTextDataset(training_frame, train_targets),
        batch_size=settings["batch_size"],
        shuffle=True,
        generator=loader_generator,
        num_workers=0,
        pin_memory=False,
    )
    valid_loader = None
    if validation_frame is not None:
        valid_loader = DataLoader(
            EncodedTextDataset(validation_frame, valid_targets),
            batch_size=settings["batch_size"],
            shuffle=False,
            num_workers=0,
            pin_memory=False,
        )

    accumulation = settings["gradient_accumulation_steps"]
    updates_per_epoch = math.ceil(len(train_loader) / accumulation)
    total_updates = updates_per_epoch * epochs
    warmup_updates = math.ceil(settings["warmup_ratio"] * total_updates)
    settings.update({
        "optimizer_updates_per_epoch": updates_per_epoch,
        "total_optimizer_updates": total_updates,
        "warmup_optimizer_updates": warmup_updates,
    })
    print(f"Training {model_id} | {len(training_frame)} rows | {epochs} epochs | {device}", flush=True)
    print(f"LR={learning_rate}, batch=8, accumulation=2, max_length=128, weight_power={weight_power}", flush=True)

    model = optimizer = scheduler = criterion = None
    best_predictions = None
    best_score = -float("inf")
    global_update = 0
    started = time.monotonic()
    try:
        # Loading the general MLM checkpoint creates a fresh six-class head.
        model = AutoModelForSequenceClassification.from_pretrained(
            model_id,
            revision=revision,
            cache_dir=str(cache_dir),
            token=False,
            num_labels=6,
            id2label=id_to_label,
            label2id=label_to_id,
            trust_remote_code=False,
            use_safetensors=True,
        ).float().to(device)
        criterion = torch.nn.CrossEntropyLoss(
            weight=torch.as_tensor(class_weights, dtype=torch.float32, device=device)
        )
        optimizer = torch.optim.AdamW(
            model.parameters(),
            lr=settings["learning_rate"],
            betas=tuple(settings["adam_betas"]),
            eps=settings["adam_epsilon"],
            weight_decay=settings["weight_decay"],
        )
        scheduler = get_linear_schedule_with_warmup(
            optimizer,
            num_warmup_steps=warmup_updates,
            num_training_steps=total_updates,
        )

        for epoch in range(1, epochs + 1):
            model.train()
            optimizer.zero_grad(set_to_none=True)
            training_loss_sum = 0.0
            training_weight_sum = 0.0
            for batch_index, batch in enumerate(train_loader):
                targets = batch.pop("labels").to(device)
                inputs = {key: value.to(device) for key, value in batch.items()}
                logits = model(**inputs).logits
                loss = criterion(logits, targets)
                batch_weight = float(criterion.weight[targets].sum().item())
                training_loss_sum += float(loss.detach().item()) * batch_weight
                training_weight_sum += batch_weight
                # Scale the final accumulation group correctly if it is shorter.
                group_start = (batch_index // accumulation) * accumulation
                group_size = min(accumulation, len(train_loader) - group_start)
                (loss / group_size).backward()
                if (batch_index + 1) % accumulation == 0 or batch_index + 1 == len(train_loader):
                    torch.nn.utils.clip_grad_norm_(model.parameters(), settings["max_grad_norm"])
                    optimizer.step()
                    scheduler.step()
                    optimizer.zero_grad(set_to_none=True)
                    global_update += 1
                    if global_update % 25 == 0 or global_update == total_updates:
                        print(
                            f"epoch {epoch}/{epochs} | optimizer step {global_update}/{total_updates} "
                            f"| train loss {training_loss_sum / training_weight_sum:.4f} "
                            f"| elapsed {time.monotonic() - started:.1f}s",
                            flush=True,
                        )
                del logits, loss, targets, inputs

            epoch_record = {
                "epoch": epoch,
                "train_loss": training_loss_sum / training_weight_sum,
                "optimizer_steps_completed": global_update,
                "elapsed_seconds": time.monotonic() - started,
            }
            predictions = None
            if valid_loader is not None:
                model.eval()
                predicted_ids = []
                probability_batches = []
                with torch.inference_mode():
                    for batch in valid_loader:
                        batch.pop("labels")
                        inputs = {key: value.to(device) for key, value in batch.items()}
                        logits = model(**inputs).logits
                        predicted_ids.extend(logits.argmax(dim=-1).cpu().tolist())
                        probability_batches.append(logits.softmax(dim=-1).cpu().numpy())
                        del logits, inputs
                probabilities = np.concatenate(probability_batches)
                assert np.isfinite(probabilities).all()
                predicted_ids = np.asarray(predicted_ids, dtype=np.int64)
                predictions = np.asarray([id_to_label[int(index)] for index in predicted_ids])
                score = float(f1_score(
                    valid_targets, predicted_ids, labels=list(range(6)),
                    average="macro", zero_division=0,
                ))
                epoch_record.update({
                    "validation_macro_f1": score,
                    "validation_accuracy": float(accuracy_score(valid_targets, predicted_ids)),
                })
                improved = score > best_score
            else:
                score = None
                improved = epoch == epochs

            result["epoch_history"].append(epoch_record)
            print(json.dumps(epoch_record, indent=2), flush=True)
            if improved:
                if score is not None:
                    best_score = score
                result["best_macro_f1"] = score
                result["best_epoch"] = epoch
                best_predictions = predictions.copy() if predictions is not None else None
                model.save_pretrained(str(output_dir), safe_serialization=True)
                tokenizer.save_pretrained(str(output_dir))
                if best_predictions is not None:
                    pd.DataFrame({
                        "post_id": validation_frame["post_id"].to_numpy(),
                        "true_label": validation_frame["label"].astype(str).to_numpy(),
                        "predicted_label": best_predictions,
                    }).to_csv(output_dir / "validation_predictions.csv", index=False)
                    np.savez_compressed(
                        output_dir / "validation_probabilities.npz",
                        probabilities=probabilities,
                        post_ids=validation_frame["post_id"].to_numpy(dtype=str),
                        labels=np.asarray(labels),
                    )
                print(f"Saved checkpoint at epoch {epoch} to {output_dir}", flush=True)
            (output_dir / "epoch_results.json").write_text(
                json.dumps(result, indent=2), encoding="utf-8"
            )

        print("Epoch scores:", flush=True)
        print(pd.DataFrame(result["epoch_history"]).to_string(index=False), flush=True)
        if best_predictions is not None:
            report = classification_report(
                validation_frame["label"].astype(str).to_numpy(),
                best_predictions,
                labels=labels,
                digits=4,
                zero_division=0,
            )
            result["best_classification_report"] = classification_report(
                validation_frame["label"].astype(str).to_numpy(),
                best_predictions,
                labels=labels,
                output_dict=True,
                zero_division=0,
            )
            print(f"Best epoch: {result['best_epoch']} | macro F1: {result['best_macro_f1']:.6f}", flush=True)
            print(report, flush=True)
        result["total_elapsed_seconds"] = time.monotonic() - started
        (output_dir / "epoch_results.json").write_text(
            json.dumps(result, indent=2), encoding="utf-8"
        )
        return result, best_predictions
    finally:
        # Return only metadata/predictions so later experiments can reclaim memory.
        model = optimizer = scheduler = criterion = None
        gc.collect()
        if device.type == "mps":
            torch.mps.empty_cache()


# Only the model changes in this first comparison with DistilRoBERTa.
ROBERTA_MODEL_ID = "FacebookAI/roberta-base"
ROBERTA_REVISION = "e2da8e2f811d1448a5b465c236feacd80ffbac7b"
assert not (set(text_groups.iloc[train_idx]) & set(text_groups.iloc[valid_idx]))
roberta_result, roberta_validation_predictions = train_candidate(
    train_part, valid_part,
    model_id=ROBERTA_MODEL_ID,
    revision=ROBERTA_REVISION,
    output_dir=experiment_dir / "roberta_base_best",
    epochs=3, seed=SEED,
)
display(pd.DataFrame([
    {"model": "DistilRoBERTa", "validation_macro_f1": distil_result["best_macro_f1"]},
    {"model": "RoBERTa-base", "validation_macro_f1": roberta_result["best_macro_f1"]},
]).sort_values("validation_macro_f1", ascending=False).round(4))
```

```python
import importlib
import shutil
import subprocess
import sys
from safetensors.torch import save_file
from transformers.utils import import_utils

DEBERTA_MODEL_ID = "microsoft/deberta-v3-base"
DEBERTA_REVISION = "8ccc9b6f36199bec6961081d44eb72fb3f7353f3"
optional_deps = experiment_dir / "optional_deps"
optional_deps.mkdir(parents=True, exist_ok=True)
optional_path = str(optional_deps.resolve())
if optional_path not in sys.path:
    sys.path.insert(0, optional_path)
importlib.invalidate_caches()
import_utils.is_sentencepiece_available.cache_clear()
import_utils.is_protobuf_available.cache_clear()
if not (import_utils.is_sentencepiece_available() and import_utils.is_protobuf_available()):
    subprocess.run([
        str(Path.home() / ".local/bin/uv"), "pip", "install",
        "--python", sys.executable,
        "--target", str(optional_deps),
        "--cache-dir", str(experiment_dir / "uv_cache"),
        "--no-deps", "--only-binary", ":all:",
        "sentencepiece==0.2.1", "protobuf==6.33.0",
    ], check=True)
    importlib.invalidate_caches()
    import_utils.is_sentencepiece_available.cache_clear()
    import_utils.is_protobuf_available.cache_clear()

deberta_snapshot = Path(snapshot_download(
    DEBERTA_MODEL_ID, revision=DEBERTA_REVISION, token=False,
    cache_dir=str(experiment_dir / "hf_cache"),
    allow_patterns=["config.json", "tokenizer_config.json", "spm.model", "pytorch_model.bin"],
    max_workers=2,
))
deberta_source = experiment_dir / "deberta_v3_base_source"
deberta_source.mkdir(parents=True, exist_ok=True)
deberta_tokenizer = AutoTokenizer.from_pretrained(
    str(deberta_snapshot), local_files_only=True,
    token=False, trust_remote_code=False,
)
assert len(deberta_tokenizer("A short emotion example.")["input_ids"]) > 2
deberta_tokenizer.save_pretrained(str(deberta_source))
shutil.copy2(deberta_snapshot / "config.json", deberta_source / "config.json")

if not (deberta_source / "model.safetensors").exists():
    deberta_state = torch.load(
        deberta_snapshot / "pytorch_model.bin", map_location="cpu", weights_only=True,
    )
    assert all(isinstance(value, torch.Tensor) for value in deberta_state.values())
    safe_state = {name: value.contiguous().clone() for name, value in deberta_state.items()}
    save_file(safe_state, str(deberta_source / "model.safetensors"), metadata={"format": "pt"})
    del deberta_state, safe_state
    gc.collect()
print("DeBERTa source ready:", DEBERTA_MODEL_ID, DEBERTA_REVISION, flush=True)


def fit_memory_transformer(
    training_frame,
    validation_frame=None,
    *,
    model_source,
    upstream_model_id,
    revision,
    output_dir,
    epochs=3,
    seed=42,
    batch_size=2,
    accumulation=8,
    max_length=128,
    lr=2e-5,
    checkpointing=True,
    class_weight_power=1.0,
):
    """Fit a local general checkpoint; select on validation or refit fixed epochs.

    Weighted losses are summed across the whole accumulation group, then its
    gradients are divided by the group's total target weight. This is exact
    group-weight normalization, unlike the earlier mean of microbatch means.
    The supplied local checkpoint and tokenizer must already be prepared.
    """
    import gc
    import json
    import math
    import platform
    import random
    import re
    import time
    from pathlib import Path

    import numpy as np
    import pandas as pd
    import sklearn
    import torch
    import transformers
    from sklearn.metrics import accuracy_score, classification_report, f1_score
    from torch.utils.data import DataLoader, Dataset
    from transformers import (
        AutoModelForSequenceClassification,
        AutoTokenizer,
        DataCollatorWithPadding,
        get_linear_schedule_with_warmup,
    )

    if not re.fullmatch(r"[0-9a-fA-F]{40}", str(revision)):
        raise ValueError("revision must be the immutable upstream commit SHA")
    for name, value in {
        "epochs": epochs, "batch_size": batch_size,
        "accumulation": accumulation, "max_length": max_length,
    }.items():
        if isinstance(value, bool) or not isinstance(value, int) or value < 1:
            raise ValueError(f"{name} must be a positive integer")
    if not np.isfinite(lr) or lr <= 0:
        raise ValueError("lr must be finite and positive")
    if not np.isfinite(class_weight_power) or class_weight_power < 0:
        raise ValueError("class_weight_power must be finite and nonnegative")
    for name, frame in (("training", training_frame), ("validation", validation_frame)):
        if frame is not None:
            if frame.empty or not {"text", "label"}.issubset(frame.columns):
                raise ValueError(f"{name} requires nonempty text and label columns")
            if frame[["text", "label"]].isna().any().any():
                raise ValueError(f"{name} contains null text or labels")
    if validation_frame is not None and "post_id" not in validation_frame:
        raise ValueError("validation_frame requires post_id for saved predictions")
    model_source = Path(model_source).expanduser().resolve()
    if not model_source.is_dir():
        raise ValueError("model_source must be an existing local checkpoint directory")
    output_dir = Path(output_dir).expanduser().resolve()
    if output_dir == model_source:
        raise ValueError("output_dir must differ from the source checkpoint")
    output_dir.mkdir(parents=True, exist_ok=True)

    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    device = torch.device("mps" if torch.backends.mps.is_available() else "cpu")
    if device.type == "mps":
        torch.mps.manual_seed(seed)
    labels = sorted(training_frame["label"].astype(str).unique().tolist())
    if len(labels) != 6:
        raise ValueError(f"Expected six classes, found {labels}")
    label_to_id = {label: index for index, label in enumerate(labels)}
    id_to_label = dict(enumerate(labels))
    train_targets = training_frame["label"].astype(str).map(label_to_id).to_numpy(dtype=np.int64)
    valid_targets = None
    if validation_frame is not None:
        mapped = validation_frame["label"].astype(str).map(label_to_id)
        if mapped.isna().any():
            raise ValueError("Validation includes an unknown class")
        valid_targets = mapped.to_numpy(dtype=np.int64)
    counts = np.bincount(train_targets, minlength=6)
    class_weights = (len(train_targets) / (6 * counts)) ** class_weight_power
    if not np.isfinite(class_weights).all() or (class_weights <= 0).any():
        raise ValueError("Class weights must be finite and positive")

    settings = {
        "seed": int(seed), "num_labels": 6, "max_length": max_length,
        "batch_size": batch_size, "gradient_accumulation_steps": accumulation,
        "nominal_effective_batch_size": batch_size * accumulation,
        "learning_rate": lr, "weight_decay": 0.01, "warmup_ratio": 0.1,
        "epochs_requested": epochs, "optimizer": "torch.optim.AdamW",
        "adam_betas": [0.9, 0.999], "adam_epsilon": 1e-8,
        "optimizer_foreach": False, "max_grad_norm": 1.0,
        "scheduler": "linear warmup followed by linear decay",
        "dtype": "float32", "amp": False, "num_workers": 0,
        "padding": "longest sequence in each batch, capped at max_length",
        "gradient_checkpointing": bool(checkpointing),
        "checkpointing_use_reentrant": False if checkpointing else None,
        "class_weight_power": float(class_weight_power),
        "class_weight_formula": "(n_training_rows / (n_classes * class_count)) ** power",
        "class_counts": dict(zip(labels, counts.astype(int).tolist())),
        "class_weights": dict(zip(labels, class_weights.astype(float).tolist())),
        "loss_normalization": "summed weighted CE / total target weight over accumulation group",
        "normalization_note": "Earlier trainer averaged normalized microbatch losses instead.",
        "selection_metric": "validation macro F1" if validation_frame is not None else None,
        "selection_rule": "highest macro F1; earliest epoch wins ties" if validation_frame is not None else "fixed final epoch",
        "trust_remote_code": False, "use_safetensors": True, "local_files_only": True,
    }
    result = {
        "model_id": str(upstream_model_id), "revision": str(revision),
        "model_source": str(model_source), "device": str(device),
        "best_macro_f1": None, "best_epoch": None, "settings": settings,
        "label_classes": labels, "train_rows": len(training_frame),
        "validation_rows": len(validation_frame) if validation_frame is not None else 0,
        "checkpoint_dir": str(output_dir), "epoch_history": [], "status": "running",
        "versions": {
            "python": platform.python_version(), "torch": str(torch.__version__),
            "transformers": transformers.__version__, "sklearn": sklearn.__version__,
            "numpy": np.__version__, "pandas": pd.__version__,
        },
        "seed_note": "Seeds are fixed; MPS operations and different microbatches can remain nondeterministic.",
    }

    def save_metadata():
        temporary = output_dir / "epoch_results.json.tmp"
        temporary.write_text(json.dumps(result, indent=2), encoding="utf-8")
        temporary.replace(output_dir / "epoch_results.json")

    started = time.monotonic()
    model = optimizer = scheduler = criterion = weight_tensor = tokenizer = None
    train_loader = valid_loader = None
    batch = inputs = targets = logits = batch_loss = parameter = None
    best_predictions = None
    best_score = -float("inf")
    update = 0
    try:
        tokenizer = AutoTokenizer.from_pretrained(
            str(model_source), local_files_only=True, trust_remote_code=False, token=False,
        )

        class TweetDataset(Dataset):
            def __init__(self, frame, targets):
                self.features = tokenizer(
                    frame["text"].astype(str).tolist(), padding=False,
                    truncation=True, max_length=max_length, return_token_type_ids=False,
                )
                self.targets = np.array(targets, dtype=np.int64, copy=True)

            def __len__(self):
                return len(self.targets)

            def __getitem__(self, index):
                return {
                    **{key: values[index] for key, values in self.features.items()},
                    "labels": int(self.targets[index]),
                }

        collator = DataCollatorWithPadding(tokenizer=tokenizer, padding=True, return_tensors="pt")
        train_loader = DataLoader(
            TweetDataset(training_frame, train_targets), batch_size=batch_size,
            shuffle=True, generator=torch.Generator().manual_seed(seed),
            collate_fn=collator, num_workers=0, pin_memory=False,
        )
        if validation_frame is not None:
            valid_loader = DataLoader(
                TweetDataset(validation_frame, valid_targets), batch_size=batch_size,
                shuffle=False, collate_fn=collator, num_workers=0, pin_memory=False,
            )
        updates_per_epoch = math.ceil(len(train_loader) / accumulation)
        total_updates = updates_per_epoch * epochs
        warmup_updates = math.ceil(0.1 * total_updates)
        settings.update({
            "optimizer_updates_per_epoch": updates_per_epoch,
            "total_optimizer_updates": total_updates,
            "warmup_optimizer_updates": warmup_updates,
        })
        print(json.dumps(result, indent=2), flush=True)
        save_metadata()
        model = AutoModelForSequenceClassification.from_pretrained(
            str(model_source), local_files_only=True, token=False,
            num_labels=6, id2label=id_to_label, label2id=label_to_id,
            trust_remote_code=False, use_safetensors=True, weights_only=True,
        ).float().to(device)
        if checkpointing:
            model.gradient_checkpointing_enable(
                gradient_checkpointing_kwargs={"use_reentrant": False}
            )
        weight_tensor = torch.tensor(class_weights, dtype=torch.float32, device=device)
        criterion = torch.nn.CrossEntropyLoss(weight=weight_tensor, reduction="sum")
        optimizer = torch.optim.AdamW(
            model.parameters(), lr=lr, betas=(0.9, 0.999), eps=1e-8,
            weight_decay=0.01, foreach=False,
        )
        scheduler = get_linear_schedule_with_warmup(
            optimizer, num_warmup_steps=warmup_updates, num_training_steps=total_updates,
        )

        for epoch in range(1, epochs + 1):
            model.train()
            optimizer.zero_grad(set_to_none=True)
            loss_total = weight_total = group_weight = 0.0
            for batch_index, batch in enumerate(train_loader):
                targets = batch.pop("labels").to(device)
                inputs = {key: value.to(device) for key, value in batch.items()}
                logits = model(**inputs).logits
                batch_loss = criterion(logits, targets)
                batch_weight = float(weight_tensor[targets].sum().item())
                loss_total += float(batch_loss.detach().item())
                weight_total += batch_weight
                group_weight += batch_weight
                batch_loss.backward()
                if (batch_index + 1) % accumulation == 0 or batch_index + 1 == len(train_loader):
                    for parameter in model.parameters():
                        if parameter.grad is not None:
                            parameter.grad.div_(group_weight)
                    torch.nn.utils.clip_grad_norm_(
                        model.parameters(), 1.0, error_if_nonfinite=True, foreach=False,
                    )
                    optimizer.step()
                    scheduler.step()
                    optimizer.zero_grad(set_to_none=True)
                    group_weight = 0.0
                    update += 1
                    if update % 25 == 0 or batch_index + 1 == len(train_loader):
                        print(
                            f"epoch {epoch}/{epochs} | optimizer step {update}/{total_updates} "
                            f"| train loss {loss_total / weight_total:.4f} "
                            f"| elapsed {time.monotonic() - started:.1f}s", flush=True,
                        )
                batch = inputs = targets = logits = batch_loss = parameter = None

            record = {
                "epoch": epoch, "train_loss": loss_total / weight_total,
                "optimizer_steps_completed": update,
            }
            probabilities = predictions = None
            score = None
            improved = epoch == epochs
            if valid_loader is not None:
                model.eval()
                chunks = []
                with torch.inference_mode():
                    for batch in valid_loader:
                        batch.pop("labels")
                        inputs = {key: value.to(device) for key, value in batch.items()}
                        logits = model(**inputs).logits
                        chunks.append(logits.softmax(dim=-1).cpu().numpy())
                        batch = inputs = logits = None
                probabilities = np.concatenate(chunks).astype(np.float32, copy=False)
                if not np.isfinite(probabilities).all():
                    raise ValueError("Non-finite validation probabilities")
                predicted_ids = probabilities.argmax(axis=1)
                predictions = np.asarray(labels)[predicted_ids]
                score = float(f1_score(
                    valid_targets, predicted_ids, labels=list(range(6)),
                    average="macro", zero_division=0,
                ))
                record.update({
                    "validation_macro_f1": score,
                    "validation_accuracy": float(accuracy_score(valid_targets, predicted_ids)),
                })
                improved = score > best_score
            record["elapsed_seconds"] = time.monotonic() - started
            result["epoch_history"].append(record)
            print(json.dumps(record, indent=2), flush=True)
            if improved:
                model.save_pretrained(str(output_dir), safe_serialization=True)
                tokenizer.save_pretrained(str(output_dir))
                result["best_macro_f1"] = score
                result["best_epoch"] = epoch
                if score is not None:
                    best_score = score
                    best_predictions = predictions.copy()
                    pd.DataFrame({
                        "post_id": validation_frame["post_id"].to_numpy(),
                        "true_label": validation_frame["label"].astype(str).to_numpy(),
                        "predicted_label": best_predictions,
                    }).to_csv(output_dir / "validation_predictions.csv", index=False)
                    np.savez_compressed(
                        output_dir / "validation_probabilities.npz",
                        post_id=validation_frame["post_id"].astype(str).to_numpy(dtype=str),
                        true_label=validation_frame["label"].astype(str).to_numpy(dtype=str),
                        label_classes=np.asarray(labels, dtype=str), probabilities=probabilities,
                    )
                    result["best_classification_report"] = classification_report(
                        validation_frame["label"].astype(str), best_predictions,
                        labels=labels, output_dict=True, zero_division=0,
                    )
                print(f"Saved checkpoint at epoch {epoch} to {output_dir}", flush=True)
            result["total_elapsed_seconds"] = time.monotonic() - started
            save_metadata()

        result["status"] = "complete"
        result["total_elapsed_seconds"] = time.monotonic() - started
        save_metadata()
        print(pd.DataFrame(result["epoch_history"]).to_string(index=False), flush=True)
        if best_predictions is not None:
            print(
                f"Best epoch: {result['best_epoch']} | macro F1: {result['best_macro_f1']:.6f}",
                flush=True,
            )
            print(classification_report(
                validation_frame["label"].astype(str), best_predictions,
                labels=labels, digits=4, zero_division=0,
            ), flush=True)
        return result, best_predictions
    except BaseException as error:
        result["status"] = "interrupted" if isinstance(error, KeyboardInterrupt) else "failed"
        result["error"] = f"{type(error).__name__}: {error}"
        result["total_elapsed_seconds"] = time.monotonic() - started
        try:
            save_metadata()
        except OSError:
            pass
        raise
    finally:
        batch = inputs = targets = logits = batch_loss = parameter = None
        model = optimizer = scheduler = criterion = weight_tensor = tokenizer = None
        train_loader = valid_loader = None
        gc.collect()
        if device.type == "mps":
            torch.mps.empty_cache()


assert not set(text_groups.iloc[train_idx]).intersection(text_groups.iloc[valid_idx])
deberta_result, deberta_validation_predictions = fit_memory_transformer(
    train_part, valid_part,
    model_source=deberta_source,
    upstream_model_id=DEBERTA_MODEL_ID,
    revision=DEBERTA_REVISION,
    output_dir=experiment_dir / "deberta_v3_base_best",
    epochs=3, seed=SEED, batch_size=2, accumulation=8,
    max_length=128, lr=2e-5, checkpointing=True, class_weight_power=1.0,
)
display(pd.DataFrame([
    {"model": "DistilRoBERTa", "validation_macro_f1": distil_result["best_macro_f1"]},
    {"model": "RoBERTa-base", "validation_macro_f1": roberta_result["best_macro_f1"]},
    {"model": "DeBERTa-v3-base", "validation_macro_f1": deberta_result["best_macro_f1"]},
]).sort_values("validation_macro_f1", ascending=False).round(4))
```

```python
# One fixed equal-weight blend, evaluated on the existing model-selection holdout.
import json
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.metrics import f1_score

blend_ids = valid_part["post_id"].astype(str).tolist()
blend_truth = valid_part["label"].astype(str).to_numpy()
blend_labels = sorted(set(blend_truth))
assert len(blend_labels) == 6 and len(set(blend_ids)) == len(blend_ids)
blend_probabilities = {}
blend_rows = []

for blend_name, id_key, label_key in [
    ("roberta_base", "post_ids", "labels"),
    ("deberta_v3_base", "post_id", "label_classes"),
]:
    blend_dir = Path(experiment_dir) / f"{blend_name}_best"
    blend_metadata = json.loads((blend_dir / "epoch_results.json").read_text())
    assert blend_metadata.get("status", "complete") == "complete"
    assert "total_elapsed_seconds" in blend_metadata
    assert len(blend_metadata["epoch_history"]) == blend_metadata["settings"]["epochs_requested"]
    assert blend_metadata["validation_rows"] == len(blend_ids)
    model_labels = blend_metadata["label_classes"]
    assert len(model_labels) == 6 and set(model_labels) == set(blend_labels)

    saved_rows = pd.read_csv(blend_dir / "validation_predictions.csv", dtype=str)
    assert saved_rows[["post_id", "true_label", "predicted_label"]].notna().all().all()
    assert saved_rows["post_id"].is_unique and set(saved_rows["post_id"]) == set(blend_ids)
    with np.load(blend_dir / "validation_probabilities.npz", allow_pickle=False) as saved:
        model_probabilities = np.array(saved["probabilities"], copy=True)
        assert saved[id_key].astype(str).tolist() == saved_rows["post_id"].tolist()
        assert saved[label_key].astype(str).tolist() == model_labels
        if "true_label" in saved.files:
            assert saved["true_label"].astype(str).tolist() == saved_rows["true_label"].tolist()
    assert model_probabilities.shape == (len(blend_ids), 6)
    assert np.isfinite(model_probabilities).all()
    assert ((model_probabilities >= 0) & (model_probabilities <= 1)).all()
    assert np.allclose(model_probabilities.sum(axis=1), 1.0, rtol=0, atol=1e-5)
    assert np.asarray(model_labels)[model_probabilities.argmax(axis=1)].tolist() == saved_rows["predicted_label"].tolist()

    row_order = pd.Index(saved_rows["post_id"]).get_indexer(blend_ids)
    column_order = [model_labels.index(label) for label in blend_labels]
    assert (row_order >= 0).all()
    assert saved_rows["true_label"].iloc[row_order].tolist() == blend_truth.tolist()
    aligned = model_probabilities[row_order][:, column_order]
    blend_probabilities[blend_name] = aligned
    individual_f1 = f1_score(
        blend_truth, np.asarray(blend_labels)[aligned.argmax(axis=1)],
        labels=blend_labels, average="macro", zero_division=0,
    )
    assert np.isclose(individual_f1, blend_metadata["best_macro_f1"], rtol=0, atol=1e-10)
    blend_rows.append({"model": blend_name, "validation_macro_f1": individual_f1})

equal_probabilities = 0.5 * blend_probabilities["roberta_base"] + 0.5 * blend_probabilities["deberta_v3_base"]
assert np.isfinite(equal_probabilities).all()
assert np.allclose(equal_probabilities.sum(axis=1), 1.0, rtol=0, atol=1e-5)
equal_blend_f1 = f1_score(
    blend_truth, np.asarray(blend_labels)[equal_probabilities.argmax(axis=1)],
    labels=blend_labels, average="macro", zero_division=0,
)
best_individual_f1 = max(row["validation_macro_f1"] for row in blend_rows)
blend_rows.append({"model": "50/50 probability average", "validation_macro_f1": equal_blend_f1})
print(pd.DataFrame(blend_rows).to_string(index=False, float_format=lambda value: f"{value:.6f}"))
print(f"Blend minus best individual: {equal_blend_f1 - best_individual_f1:+.6f}")
print("This reuses the model-selection holdout. A small gain alone does not establish better generalization.")
```
